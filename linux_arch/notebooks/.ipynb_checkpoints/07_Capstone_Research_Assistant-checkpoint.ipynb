# Notebook 07 — Capstone: a research assistant that uses everything

**Time:** about 60 minutes

**What you will build** — an assistant that answers a research question from our local documents:

| Step | Pattern from the workshop | Who does it |
|---|---|---|
| 1. Split the question into 3 sub-questions | Planning (05) | Model |
| 2. Search the documents for each sub-question | Tool use (03) | **Code** |
| 3. Write notes with sources | Multi-agent roles (06) | Model |
| 4. Write the report | Workflow (01) | Model |
| 5. Check sources and sections, fix once | Reflection + evaluation (02, 04) | **Code** checks, model fixes |
| 6. Add the reference list and save | — | **Code** |

The reference list is built **by code** from documents we really used, so the report cannot invent sources.

### ▶️ Run the two setup cells below first
They check that Ollama is ready and create `ask()`, the one helper we use to talk to the model.

```python
# STEP 0 — Is Ollama running? Is the model downloaded?
import ollama

MODEL = "llama3.2"     # local model (bigger one: "qwen2.5:7b")

try:
    installed = [m.model for m in ollama.list().models]
    print("Models on this computer:", installed)
    if MODEL in installed or MODEL + ":latest" in installed:
        print("✅ Ready to go with", MODEL)
    else:
        print(f"⚠️ {MODEL} is missing. Run: ollama pull {MODEL}")
except ConnectionError:
    print("⚠️ Ollama is not running. Start it, then run this again:")
    print("   Arch Linux: sudo systemctl start ollama")
    print("   Windows 11: open Ollama from the Start menu")
```

```python
# STEP 1 — A helper: send a message, get the answer back as text
DEFAULT_ROLE = "You are a helpful assistant. Keep answers short."

def ask(prompt, system=DEFAULT_ROLE):
    reply = ollama.chat(
        model=MODEL,
        messages=[{"role": "system", "content": system},
                  {"role": "user", "content": prompt}],
        options={"temperature": 0},    # 0 = more predictable
    )
    return reply.message.content

print("ask() is ready.")
```

## Step 0 — The question and the documents

```python
import json, re
from pathlib import Path
from pydantic import BaseModel

QUESTION = "How can AI improve services for university students?"

path = "../datasets/research_documents.json"
documents = json.load(open(path, encoding="utf-8"))["documents"]
DOCS_BY_ID = {d["doc_id"]: d for d in documents}

WARNING_SIGNS = ["ignore all previous instructions",
                 "ignore previous instructions",
                 "api key", "cite this"]

def looks_suspicious(text):
    text = text.lower()
    for sign in WARNING_SIGNS:
        if sign in text:
            return True
    return False

print(len(documents), "documents loaded")
```

## Step 1 — Plan: split the question into sub-questions

```python
class SubQuestions(BaseModel):
    questions: list[str]

prompt = ("Split this research question into exactly 3 short "
          "sub-questions:\n" + QUESTION)
reply = ollama.chat(model=MODEL,
                    messages=[{"role": "user", "content": prompt}],
                    format=SubQuestions.model_json_schema(),
                    options={"temperature": 0})
try:
    result = SubQuestions.model_validate_json(reply.message.content)
    sub_questions = result.questions[:3]
except Exception:
    sub_questions = []
if not sub_questions:          # plan failed → use the main question
    sub_questions = [QUESTION]

for q in sub_questions:
    print("•", q)
```

## Step 2 — Search: find documents for each sub-question (a simple keyword search, no AI)

```python
def search(query, top=3):
    words = re.findall(r"[a-z]+", query.lower())
    query_words = [w for w in words if len(w) > 3]
    scored = []
    for doc in documents:
        if looks_suspicious(doc["text"]):
            continue                # never use suspicious documents
        text = (doc["title"] + " " + doc["text"]).lower()
        score = 0
        for word in query_words:
            if word in text:
                score += 1
        if score > 0:
            scored.append((score, doc["doc_id"]))
    scored.sort(reverse=True)       # highest score first
    return [doc_id for score, doc_id in scored[:top]]
```

```python
found = {}               # sub-question → list of document ids
for q in sub_questions:
    found[q] = search(q + " " + QUESTION)
    print(q, "→", found[q])
```

## Step 3 — Notes: the researcher writes facts with sources

```python
RESEARCHER = ("You are a careful researcher. Write 2 or 3 short "
              "facts that answer the sub-question. After each "
              "fact put its source id, like [DOC-01]. Use ONLY "
              "the documents given.")

notes = []
for q, ids in found.items():
    if not ids:
        notes.append(f"## {q}\nNo documents found.")
        continue
    docs_text = ""
    for i in ids:
        docs_text += f"[{i}] {DOCS_BY_ID[i]['text']}\n\n"
    facts = ask(f"Sub-question: {q}\n\nDocuments:\n{docs_text}",
                system=RESEARCHER)
    notes.append(f"## {q}\n{facts}")
    print(notes[-1], "\n")

all_notes = "\n\n".join(notes)
```

## Step 4 — Write the report

```python
WRITER = ("You are a precise writer. Write a short report with "
          "exactly these headings: '## Introduction', "
          "'## Findings', '## Conclusion'. Use only facts from "
          "the notes and keep their [DOC-xx] sources. "
          "Do not add a reference list.")

request = f"Research question: {QUESTION}\n\nNotes:\n{all_notes}"
report = ask(request, system=WRITER)
print(report)
```

## Step 5 — Check with code, and fix once if needed

```python
USED_IDS = set()             # every document id we retrieved
for ids in found.values():
    USED_IDS.update(ids)

HEADINGS = ["## Introduction", "## Findings", "## Conclusion"]

def check_report(text):
    problems = []
    for heading in HEADINGS:
        if heading.lower() not in text.lower():
            problems.append(f"missing heading {heading}")
    cited = set(re.findall(r"DOC-\d+", text))
    if not cited:
        problems.append("no sources are cited")
    made_up = cited - USED_IDS
    if made_up:
        problems.append(f"unknown sources: {sorted(made_up)}")
    return problems
```

```python
problems = check_report(report)
print("Problems:", problems or "none ✅")

if problems:                          # one revision only
    report = ask(f"Fix these problems: {problems}\n\n"
                 f"Notes:\n{all_notes}\n\nReport:\n{report}",
                 system=WRITER)
    problems = check_report(report)
    print("After the fix:", problems or "none ✅")
```

## Step 6 — Add the references (by code) and save the report

```python
cited = sorted(set(re.findall(r"DOC-\d+", report)) & USED_IDS)
references = ""
for i in cited:
    references += f"- [{i}] {DOCS_BY_ID[i]['title']}\n"

final_report = (f"# {QUESTION}\n\n{report}\n\n"
                f"## References\n{references or '- (none)'}\n")
if problems:
    final_report += f"\n> ⚠️ Checks still found: {problems}\n"

Path("../outputs").mkdir(exist_ok=True)
out = Path("../outputs/research_report.md")
out.write_text(final_report, encoding="utf-8")
print(final_report)
print("📄 Saved to outputs/research_report.md")
```
